```python
input_img = "/home/Data/FOMO-300K/derivatives/nextbrain/PT014_CoRR/sub-0001/ses-01/anat/seg.right.nii.gz"
# input_img = "../data/test/changed_recon_holes_0.2mm_L1/case-1.nii.gz"
output_dir = "../data/train/corrupted_images"
os.makedirs(output_dir, exist_ok=True)

base = os.path.basename(input_img).replace(".nii.gz", "")
for i, (img, params) in enumerate(corrupt_nifti(input_img, (-20, -5), "coronal", n=2, seed=0)):
    # params has: axis, tilt_about, angle_deg, normal_ras, thickness_mm, phase_mm,
    #             changed_fraction, events (list of applied corruptions)
    stem = f"{output_dir}/{base}_corrupt_{i:03d}"
    nib.save(img, stem + ".nii.gz")
    with open(stem + ".json", "w") as f:
        json.dump(params, f, indent=2)
    print(f"[{i}] angle={params['angle_deg']:.2f} deg  changed={100 * params['changed_fraction']:.2f}%  "
          f"events={len(params['events'])}")
```
